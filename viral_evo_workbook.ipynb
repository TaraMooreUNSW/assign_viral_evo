````python
# get ref seq (NC_045512.2)
REF_SEQ_ID = "NC_045512.2"

# set up ref seq variables
ref_seq = None
ref_seq_collection_time = None
ref_seq_collection_loc = None

# {id = (seq, location, time, days)}
dependents = {}

# get file seqs
for variant in SeqIO.parse("data/p1-sars-cov-2-variants.fasta", "fasta"):
    var_id, var_loc, var_time = variant.id.split('|')

    var_seq = variant.seq

    # find ref seq
    if var_id == REF_SEQ_ID:
        ref_seq = var_seq
        ref_seq_collection_time = pd.to_datetime(var_time)
        ref_seq_collection_loc = var_loc

    dependents[var_id] = (var_seq, var_loc, pd.to_datetime(var_time))

# set days since ref_seq
for entry_id, e_info in dependents.items():
    _, _, e_time = e_info

    days_since_start = (pd.to_datetime(e_time) - ref_seq_collection_time) / pd.Timedelta(days=1)
    dependents[entry_id] += (int(days_since_start),)

# sort by var_time from earliest to most recent
sorted_dependents = dict(sorted(dependents.items(), key=lambda time: time[1][3]))

# Calculate the JC corrected genetic distance from the reference sequence to all other and plot its dependence on the time elapsed from this starting point.
# { id: (jc_dist, days_since_start) }
genetic_dist_from_ref = {}

for d_id in sorted_dependents:
    d_seq, _, _, d_days= sorted_dependents[d_id]

    genetic_dist_from_ref[d_id] = (jukes_cantor(ref_seq, d_seq), d_days)

# Create a scatterplot of viruses
# genetic distance on the y-axis and time in days (the number of days since the starting point) on your x-axis.
fig, ax = plt.subplots(figsize=(20,8))

# plot
x_values = []
y_values = []

# plot all seqs
for label, coords in genetic_dist_from_ref.items():
    score, days = coords

    if label == REF_SEQ_ID:
        # plot reference
        ax.scatter(days, score, label=REF_SEQ_ID, color="black", marker="x", zorder=5)
    else:
        ax.scatter(days, score, alpha=0.75, s=50)

    x_values.append(days)
    y_values.append(score)

# Estimate the mutation rate using linear regression and overlay your plot with the regression line.
# using reference in linear regression
slope, intercept = np.polyfit(x_values, y_values, deg=1) # slope = mutation rate per genome per day
y_trend = slope * np.array(x_values) + intercept

# mutation rate per nt per day
exclude = {"-", "N"}
sum_seq_len = sum(1 for char in ref_seq if char not in exclude)
for id in dependents:
    sum_seq_len += sum(1 for char in dependents[id][0] if char not in exclude)

mean_seq_length = int(sum_seq_len / len(dependents) + 1)
mutation_rate_per_nt_per_day = slope / mean_seq_length
print(f"Mutation rate per genome per day: {round(slope, 3)}")
print(f"Mutation rate per nt per day (using mean seq length): {mutation_rate_per_nt_per_day}")


ax.plot(x_values, y_trend, label=f"y = {slope:.3f}x  {intercept:+.3f}")
ax.legend(loc="center left")

plt.tight_layout()
plt.savefig('problem1.svg')
````

## Problem 3: Reinterpreting mutation rates

Majority of RNA viral genomes are single-stranded RNA, where only one strand of the viruses encode for the genes. We divide these viruses into positive-sense and negative-sense viruses, based on where they code for genes.
Positive-sense RNA directly codes for genes and acts like mRNA, meaning it can be directly used by the host's ribosomes to make proteins.
Negative-sense RNA is a reverse complement of the positive-sense RNA and does not code for genes directly. It must first be converted into positive-sense RNA by the virus's own enzymes before it can be used to make proteins.
Positive-sense RNA viruses mutate faster because their replication process is simpler and mostly lacks error correction. Negative-sense RNA viruses, on the other hand, mutate more slowly since they have an extra step in replication and require more functional genes to support replication.

## Problem 4: Variant-specific mutations

As the virus mutates, it inevitably evolves and proliferates around the world. Every so often, some mutations may prove especially beneficial to the spread of the virus, and this version of the virus spreads faster than other versions. When a version of a virus becomes especially prevalent inside a population, we call this a virus *variant*. Variants are nothing more than a naming scheme for viruses that have specific mutations. For instance, in Slovenia during 2022, we were dealing with the Omicron variant. Think of this as observing natural selection in real time. Some viruses have mutations that enable them to spread more easily throughout our population, which inevitably leads to the demise of other virus variants, which are not as good at proliferation. The result is survival of the fittest at the viral level, where, unfortunately, the fittest viruses seem to cause the most damage to us humans.

How do we identify variants? A variant is determined by several so-called *defining mutations*. Mutations can either be synonymous or nonsynonymous. Synonymous mutations are changes in nucleotide bases that result in the same encoded amino acid and are thus less important. Nonsynonymous mutations are nucleotide mutations that alter the amino acid sequence of a protein.

````python
REF_SEQ_ID = "NC_045512.2"
ref_seq = None


# Extract the Alpha and Delta variants according to the indices in lists `alpha_variants` and `delta_variants`, respectively.
seqs = {} # id: (seq, variant_type)
for s in SeqIO.parse("data/p1-sars-cov-2-variants.fasta", "fasta"):
    s_id, _, _ = s.description.split('|')
    s_seq = s.seq

    if s_id == REF_SEQ_ID:
        ref_seq = s_seq
    else:
        if s_id in alpha_variants or delta_variants:
            variant_type = "Alpha" if s_id in alpha_variants else "Delta"
            seqs[s_id] = (s_seq, variant_type)


# Calculate the nucleotide mismatches between each sequence and the reference sequence for the variant.
# exlucde indels and N bases
exclude = {"-", "N"}
# mismatches = { id: {'mismatches': no. mismatches, 'depth': total seqs searched, 'bases': set of bases seen that don't match ref seq base}
mismatches = {}

# mismatches at nucleotide position
mismatch_nt_position_alpha = {idx: {"mismatches": 0, "depth": 0, "bases": set()} for idx, _ in enumerate(ref_seq)}
mismatch_nt_position_delta = {idx: {"mismatches": 0, "depth": 0, "bases": set()} for idx, _ in enumerate(ref_seq)}

for seq_id, seq_info in seqs.items():
    s_seq, var_type = seq_info
    # ignore '-'  and 'N
    valid_seq_pairs = [(index, a, b) for index, (a, b) in enumerate(zip(ref_seq, s_seq)) if a not in exclude and b not in exclude]
    no_mismatches = sum(a != b for _, a, b in valid_seq_pairs)

    # add to count of number of mismatches at index of reference str
    # keeping each variant separate
    for pair_info in valid_seq_pairs:
        index, ref_pair, seq_pair = pair_info
        if var_type == "Alpha":
            mismatch_nt_position_alpha[index]["mismatches"] += 1 if ref_pair != seq_pair else 0
            mismatch_nt_position_alpha[index]["depth"] += 1
            mismatch_nt_position_alpha[index]["bases"].update((seq_pair,) if ref_pair != seq_pair else ())
        else:
            mismatch_nt_position_delta[index]["mismatches"] += 1 if ref_pair != seq_pair else 0
            mismatch_nt_position_delta[index]["depth"] += 1
            # mismatch_nt_position_delta[index]["bases"].add(seq_pair)
            mismatch_nt_position_delta[index]["bases"].update((seq_pair,) if ref_pair != seq_pair else ())

    mismatches[seq_id] = no_mismatches

# compute the average mismatch ratio for each nucleotide position to generate an array of mutation frequencies ranging from 0 to 1.
# A value of 0 means no mutations occurred at that position in any sequence,
# while a value of 1 indicates that all sequences have a mutation at that position relative to the reference.
# Ensure that mutations are calculated separately for each variant.

# mutation_frequencies = [] # index is position within ref seq
mutation_frequencies_alpha = [] # index is position within ref seq
mutation_frequencies_delta = [] # index is position within ref seq

for position, mismatch_dict in mismatch_nt_position_alpha.items():
    depth = mismatch_dict["depth"]
    no_mismatches = mismatch_dict["mismatches"]
    if depth == 0:
        mutation_frequencies_alpha.append(0.0)
    else:
        mutation_frequencies_alpha.append(no_mismatches / depth)

for position, mismatch_dict in mismatch_nt_position_delta.items():
    depth = mismatch_dict["depth"]
    no_mismatches = mismatch_dict["mismatches"]
    if depth == 0:
        mutation_frequencies_delta.append(0.0)
    else:
        mutation_frequencies_delta.append(no_mismatches / depth)

# Plot the mutation occurrences across the whole genome for the Alpha and Delta variants separately on one figure (two subplots, one on top of another). 
# x-axis is sequence index and y-axis is the average mutation occurance. 
# Show only the part of the genome above the 20,000 nucleotides
position_start = 20000

fig, (ax1, ax2) = plt.subplots(2,1, figsize=(12,8), sharey=True)

x_axis = len(ref_seq)

ax1.plot(range(position_start, x_axis), mutation_frequencies_alpha[position_start:])
ax1.set_title("Alpha variants")
ax1.set_ylabel("Mutation frequency")
ax1.set_xlabel("Sequence index")

ax2.plot(range(position_start, x_axis), mutation_frequencies_delta[position_start:])
ax2.set_title("Delta variants")
ax2.set_ylabel("Mutation frequency")
ax2.set_xlabel("Sequence index")

cmap = plt.get_cmap('tab20b').colors[2::4]
location_colours = {name: cmap[i] for i, name in enumerate(gene_locations)}

# and mark locations of the SARS-CoV-2 genes (with colors and labels).
for name, (start, stop) in gene_locations.items():
    ax1.axvspan(start,stop, color=location_colours[name], alpha=0.3)
    mid_x = (start + stop) /2
    ax1.text(x=mid_x, y=1.1, s=name, ha="center", va="top")

    ax2.axvspan(start,stop, color=location_colours[name], alpha=0.3)
    ax2.text(x=mid_x, y=1.1, s=name, ha="center", va="top")

min_y, _ = ax1.get_ylim()
ax1.set_ylim(min_y,1.15)
ax2.set_ylim(min_y,1.15)


# We say a mutation is vital if its occurrence is higher than 0.5.
# Find all vital mutations in the spike gene ("S") and compare results between variants.
# There are a few sites where both variants mutated, but only one where the same mutation occurred.

(s_start, s_stop) = gene_locations["S"]

vital_mut_alpha = [] # position in ref seq
vital_mut_delta = []

for pos, freq in enumerate(mutation_frequencies_alpha[s_start:s_stop], start=s_start):
    if freq > 0.5:
        vital_mut_alpha.append(pos)

for pos, freq in enumerate(mutation_frequencies_delta[s_start:s_stop], start=s_start):
    if freq > 0.5:
        vital_mut_delta.append(pos)

# Find indices (based on ref genome) of each site on the S gene where mutations occurred in both variants.
vital_mut_both = list(set(vital_mut_alpha) & set(vital_mut_delta))
print(vital_mut_both)

# Store your plot in the `problem3.svg` and make it visually appealing.
plt.tight_layout()
plt.savefig('problem3.svg')

# Store the only same mutation into the `same_mutation` variable as a string denoting reference nucleotide, mutation site, and the variant nucleotide
same_mutation_str = ""
for vital_mut in vital_mut_both:
    ref_base = ref_seq[vital_mut]

    mismatch_info_alpha = mismatch_nt_position_alpha[vital_mut]
    mismatch_bases_alpha = mismatch_info_alpha["bases"]

    mismatch_info_delta = mismatch_nt_position_delta[vital_mut]
    mismatch_bases_delta = mismatch_info_delta["bases"]

    shared_mut = set(mismatch_bases_alpha) & set(mismatch_bases_delta)
    mutation = {mut for mut in shared_mut if mut != ref_base}

    if mutation:
        mut_base = next(iter(mutation))
        same_mutation_str = f"{ref_base}{vital_mut}{mut_base}"
        break

print(same_mutation_str)
````

````python
# You have 665 pre-aligned SARS-CoV-2 spike protein sequences collected from Australia at various time points (`data/p5-aus-spike-proteins.fasta`). 
# Your job is to determine the variant of each spike protein sequence. We have also prepared a JSON file (*variants.json*) containing all the information on SARS-CoV-2 variants that you'll need. 
# Use the variant names and nonsynonymous mutations defined in this file for your classification. We have removed some of the variants for the purposes of this assignment.

# Your task is to look through each of the provided SARS-Cov-2 sequences and assign it to a variant.
# To assign a variant, count the number of corresponding mutations on the spike protein sequence. Note that mutation locations start with index 1 in the _variants.json_ file.

# spike_seqs = {'accession_id': {protein seq, variant: from json file}}
spike_seqs = {}

for spike in SeqIO.parse("data/p5-aus-spike-proteins.fasta", "fasta"):
    _, spike_date, spike_id = spike.id.split('|')
    spike_seq = spike.seq

    spike_seqs[spike_id] = {"seq": spike_seq, "date": spike_date}

with open("data/variants.json", encoding="utf-8") as var_file:
    data = json.load(var_file)

# only consider non-synonomous mutations
for s_id, s_info in spike_seqs.items():
    # print(f"{s_id}:")

    variant_percentage = 0
    variant_muts = 0
    variant_match = ""

    for variant, v_type in data.items():
        mut_matches = 0
        # maybe filter out all genes not S, E, N, M for variant typing
        spike_protein_genes = ["S", "E", "M", "N"]
        # no_mutations = len(v_type["nonsynonymous"])
        no_mutations = sum(1 for info in v_type["nonsynonymous"] if info.get("gene") in spike_protein_genes)
        # no_mutations = sum(1 for info in v_type["nonsynonymous"] if info.get("gene") == "S")

        for mutation in v_type["nonsynonymous"]:
            # Note that mutation locations start with index 1 in the _variants.json_ file.
            gene = mutation["gene"]
            expect_left = mutation["left"]
            expect_right = mutation["right"]
            position = mutation["pos"]

            if gene not in spike_protein_genes:
            # if gene != "S":
                continue

            # check for insertions and move to next mutation if found
            if expect_left == '-':
                continue

            # check mut not out of range:
            if position <= len(s_info["seq"]):
                # calculate matches
                seq_base = s_info["seq"][position - 1]

                mut_matches += 1 if expect_right == seq_base and seq_base != 'X' else 0

        curr_var_percent = mut_matches / no_mutations
        # print(f"{variant} has no. {mut_matches} mut matches vs total possible mutations: {no_mutations}. Freq {curr_var_percent}")

        if curr_var_percent >= variant_percentage:
            variant_percentage = curr_var_percent
            variant_muts = mut_matches
            variant_match = variant

    # print(f'curr best match: {variant_match} at {variant_percentage * 100}%')

    # Assign the variant to a sequence with the highest percentage of matching mutations to a given variant. 
    # If you do not get at least 51% match for any of the variants, you should assign the genome to the *UNKNOWN* class. 
    # If a sequence has no mutations throughout the whole genome compared to the reference, assign it to the *NO_MUTATIONS* class. 
    if variant_muts > 0 and variant_percentage < 0.51:
        spike_seqs[s_id]["variant"] = "UNKNOWN"
    elif variant_muts == 0:
        spike_seqs[s_id]["variant"] = "NO_MUTATIONS"
    else:
        spike_seqs[s_id]["variant"] = variant_match


# Save your answers to `problem5-classification.csv`. This file should contain two columns.
# The first column indicates accession id,
# the second column should indicates the variant display name
with open('problem5-classification.csv', 'w', newline='', encoding='utf-8') as var_csv:
    writer = csv.writer(var_csv)

    for seq_id, seq_info in spike_seqs.items():
        row = [seq_id, seq_info["variant"]]
        writer.writerow(row)


sorted_spike_seqs = dict(sorted(
    spike_seqs.items(),
    key=lambda info: datetime.strptime(info[1]['date'], "%Y-%m-%d")
))
# print(sorted_spike_seqs)


# After you have assigned all the sequences to a variant,
# create a plot showing the change in distribution over time. 
# More concretely, we have given you up to 30 sequences per month. 
# Calculate the percentage of each variant in each month, and plot the distribution over time. 

df = pd.DataFrame(sorted_spike_seqs)
df = df.T
df["date"] = pd.to_datetime(df['date'])

# grouped = df.groupby(pd.Grouper(key="date", freq="ME"))
grouper = pd.Grouper(key="date", freq="ME")
variant_proportion = df.groupby(grouper)['variant'].value_counts(normalize=True).reset_index(name="var_proportion")
pivot_df = variant_proportion.pivot(index="date", columns="variant", values="var_proportion").fillna(0)

# Your plots should resemble https://covariants.org/per-country, which you can use to validate your work. 
# Note that your plot will probably have a lot of *UNKNOWN* viral assignments.
# use plt.stackplot
# x-axis: month, y-axis: proportion

plt.figure(figsize=(16, 8))
x_axis = pivot_df.index.strftime('%b %Y')
y_axis = [pivot_df[col] for col in pivot_df.columns]


cmap = plt.get_cmap('tab20')
variant_colours = [cmap(i) for i in np.linspace(0,1, len(pivot_df.columns))]


plt.stackplot(x_axis, y_axis, labels=pivot_df.columns, alpha=0.75, colors=variant_colours)
plt.xticks(rotation=90)
plt.legend(
    title="SARS-CoV-2 variants",
    loc="right",
    bbox_to_anchor=(1.15,0.75)
)

# Save your figure into `problem5.svg`.
plt.tight_layout()
plt.savefig('problem5.svg')
````
